```python
# ================= USER BEHAVIOR =================

oct_user = pd.crosstab(
    oct_df["user_id"],
    oct_df["event_type"]
).fillna(0)

nov_user = pd.crosstab(
    nov_df["user_id"],
    nov_df["event_type"]
).fillna(0)

print("OCTOBER")
print("Users:", len(oct_user))
print("Users with cart:", (oct_user["cart"] > 0).sum())
print("Users with purchase:", (oct_user["purchase"] > 0).sum())

print("\nNOVEMBER")
print("Users:", len(nov_user))
print("Users with cart:", (nov_user["cart"] > 0).sum())
print("Users with purchase:", (nov_user["purchase"] > 0).sum())

print("\nEvents per user - October")
print(oct_user[["view", "cart", "purchase"]].describe())

print("\nEvents per user - November")
print(nov_user[["view", "cart", "purchase"]].describe())
```

```python
# ================= USER FUNNEL =================

def funnel_analysis(user_df, month):

    view_users = (user_df["view"] > 0).sum()
    cart_users = (user_df["cart"] > 0).sum()
    purchase_users = (user_df["purchase"] > 0).sum()

    print(f"\n========== {month} ==========")

    print("View users:", view_users)
    print("Cart users:", cart_users)
    print("Purchase users:", purchase_users)

    print(
        "View → Cart:",
        round(cart_users / view_users * 100, 2), "%"
    )

    print(
        "Cart → Purchase:",
        round(purchase_users / cart_users * 100, 2), "%"
    )

    print(
        "View → Purchase:",
        round(purchase_users / view_users * 100, 2), "%"
    )


funnel_analysis(oct_user, "OCTOBER")
funnel_analysis(nov_user, "NOVEMBER")
```

```python
# ================= HIGH-ENGAGEMENT CANDIDATES =================

oct_candidates = oct_user[
    (oct_user["view"] >= 5) &
    (oct_user["cart"] > 0) &
    (oct_user["purchase"] == 0)
]

nov_candidates = nov_user[
    (nov_user["view"] >= 5) &
    (nov_user["cart"] > 0) &
    (nov_user["purchase"] == 0)
]

print("October candidates:", len(oct_candidates))
print("November candidates:", len(nov_candidates))
```

```python
# ================= VIEW COUNT vs PURCHASE =================

oct_view_purchase = oct_user.copy()

oct_view_purchase["view_bucket"] = pd.cut(
    oct_view_purchase["view"],
    bins=[0, 1, 2, 3, 5, 10, 20, 50, 100, float("inf")],
    labels=[
        "1",
        "2",
        "3",
        "4-5",
        "6-10",
        "11-20",
        "21-50",
        "51-100",
        "100+"
    ]
)

threshold_analysis = (
    oct_view_purchase
    .groupby("view_bucket", observed=True)
    .agg(
        users=("purchase", "size"),
        purchasers=("purchase", lambda x: (x > 0).sum())
    )
)

threshold_analysis["purchase_rate"] = (
    threshold_analysis["purchasers"]
    / threshold_analysis["users"]
    * 100
)

print(threshold_analysis)
```

```python
# ================= PRODUCT FUNNEL =================

oct_product = pd.crosstab(
    oct_df["product_id"],
    oct_df["event_type"]
).fillna(0)

oct_product["view_to_cart_rate"] = (
    oct_product["cart"] /
    oct_product["view"].replace(0, pd.NA)
    * 100
)

oct_product["cart_to_purchase_rate"] = (
    oct_product["purchase"] /
    oct_product["cart"].replace(0, pd.NA)
    * 100
)

oct_product["view_to_purchase_rate"] = (
    oct_product["purchase"] /
    oct_product["view"].replace(0, pd.NA)
    * 100
)

print(oct_product.sort_values(
    "view",
    ascending=False
).head(20))
```

```python
import pandas as pd

def check_data_types(df, name):

    print(f"\n========== {name} ==========")

    # event_time
    invalid_event_time = pd.to_datetime(
        df["event_time"],
        errors="coerce",
        utc=True
    ).isna().sum()

    # event_type
    invalid_event_type = (
        ~df["event_type"].dropna().isin(
            ["view", "cart", "purchase"]
        )
    ).sum()

    # integer columns
    invalid_product_id = pd.to_numeric(
        df["product_id"],
        errors="coerce"
    ).isna().sum()

    invalid_category_id = pd.to_numeric(
        df["category_id"],
        errors="coerce"
    ).isna().sum()

    invalid_user_id = pd.to_numeric(
        df["user_id"],
        errors="coerce"
    ).isna().sum()

    # numeric price
    invalid_price = pd.to_numeric(
        df["price"],
        errors="coerce"
    ).isna().sum()

    # string columns
    invalid_category_code = (
        df["category_code"].notna()
        & ~df["category_code"].map(lambda x: isinstance(x, str))
    ).sum()

    invalid_brand = (
        df["brand"].notna()
        & ~df["brand"].map(lambda x: isinstance(x, str))
    ).sum()

    invalid_session = (
        df["user_session"].notna()
        & ~df["user_session"].map(lambda x: isinstance(x, str))
    ).sum()

    results = {
        "event_time": invalid_event_time,
        "event_type": invalid_event_type,
        "product_id": invalid_product_id,
        "category_id": invalid_category_id,
        "category_code": invalid_category_code,
        "brand": invalid_brand,
        "price": invalid_price,
        "user_id": invalid_user_id,
        "user_session": invalid_session
    }

    print("\nInvalid values:")
    for column, count in results.items():
        print(f"{column:20} {count:,}")

    return results


oct_type_check = check_data_types(oct_df, "OCTOBER")
nov_type_check = check_data_types(nov_df, "NOVEMBER")
```
